1. Create a numpy array of size 2 x 2

```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print (array_2x2)
```

2. Create a random numpy array of size 3 x 4

```python
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones:")
print(ones_array)

print("Array of Zeros:")
print(zeros_array)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
rand_arr = np.random.rand(6,3)
print(rand_arr)

```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n,p,m = 4,3,5
array_np = np.random.rand(n,p)
array_pm = np.random.rand(p,n)
result_dot = array_np @ array_pm
print("Array 1 Shape: ", array_np.shape)
print("Array 2 Shape: ", array_pm.shape)
print("\nResulting Dot Product Shape:" , result_dot.shape)
print("\nDot Product Result: \n", result_dot)
```
